```python
# extrator_texto

# Configurar o logger
logging.basicConfig(
    level=logging.INFO,
    format='[%(levelname)s] %(message)s'
)

def to_pdf(img_path, caminho_pdf):
    with open(caminho_pdf, "wb") as f:
        f.write(img2pdf.convert(img_path))

def ler_texto_pdf(caminho_pdf: str) -> str:
    """
    Lê todas as páginas de um PDF e retorna o conteúdo concatenado em uma única string.
    
    Args:
        caminho_pdf (str): Caminho para o arquivo PDF.
    
    Returns:
        str: Texto completo do PDF.
    """
    caminho = Path(caminho_pdf)
    if not caminho.exists():
        raise FileNotFoundError(f"Arquivo não encontrado: {caminho_pdf}")

    texto = []
    try:
        with fitz.open(caminho_pdf) as doc:
            for pagina in doc:
                texto.append(pagina.get_text())
    except Exception as e:
        raise RuntimeError(f"Erro ao ler o PDF: {e}")

    return "\n".join(texto)
```

```python
#import easyocr
import re
import os

#reader = easyocr.Reader(['pt'])

def filtrar_valor(valor_str):
    try:
        valor = float(valor_str.replace(".", "").replace(",", "."))
        if 2 <= valor <= 300:  # só valores dentro do intervalo
            return valor_str
    except ValueError:
        pass
    return None

from paddleocr import PaddleOCR
ocr = PaddleOCR(lang='pt')  # inicializa uma vez

def ler_texto_img(imagem):
    results = ocr.ocr(imagem, cls=True)
    string = ' '.join([res[1][0] for linha in results for res in linha])

    # --- Datas ---
    match_datas = re.findall(r'\d{2}/\d{2}/\d{4}', string)
    datas_validas = {
        d for d in match_datas
        if d.endswith("2025") and d[3:5] == "07"  # julho de 2025
    }

    # --- Valores ---
    match_valores = re.findall(r'\d{1,3}(?:\.\d{3})*,\d{2}', string)
    valores_validos = {
        v for v in match_valores if filtrar_valor(v)
    }

    datas_final = list(datas_validas) if datas_validas else ["Não encontrada"]
    valores_final = list(valores_validos) if valores_validos else ["Não encontrado"]

    print(f"Datas: {datas_final}, Valores: {valores_final}")
    # return datas_final, valores_final

def carregar_arquivos_da_pasta(pasta, extensao=".jpg"):
    return [
        os.path.abspath(os.path.join(pasta, f)).replace("\\", "/")
        for f in os.listdir(pasta)
        if f.endswith(extensao)
    ]
```
